```python
from google.colab import drive
drive.mount("/content/drive")

import os, shutil, subprocess

PROJECT_DIR = "/content/drive/MyDrive/Concrete_defects_CNN_v2"
DATA_DIR = os.path.join(PROJECT_DIR, "Data")
DATASET_DIR = os.path.join(DATA_DIR, "dacl1k_raw")
os.makedirs(DATA_DIR, exist_ok=True)

print("PROJECT_DIR:", PROJECT_DIR)
print("DATASET_DIR (final home in your Drive):", DATASET_DIR)
```

```python
if os.path.exists(DATASET_DIR) and any(os.scandir(DATASET_DIR)):
    print(f"Already saved in your Drive at {DATASET_DIR} — nothing to do.")
else:
    from bikit.utils import download_dataset
    download_dataset("dacl1k")

    # Confirmed actual location from a real run — check this first
    known_path = "/root/.cache/bikit/dacl1k"

    if os.path.exists(known_path):
        source_path = known_path
    else:
        # Fallback: search if the known path doesn't exist in your bikit version
        result = subprocess.run(
            ["find", "/content", "/root", "-iname", "*dacl1k*", "-maxdepth", "4"],
            capture_output=True, text=True
        )
        print("Known path not found. Candidates:\n", result.stdout)
        source_path = None

    if source_path:
        shutil.copytree(source_path, DATASET_DIR)
        print(f"Copied dataset to your Drive: {DATASET_DIR}")
    else:
        raise RuntimeError("Could not locate downloaded dataset — check the candidates printed above "
                            "and set source_path manually, then re-run this cell.")
```

```python
print("Saved dataset contents:")
for f in sorted(os.listdir(DATASET_DIR)):
    print(" ", f)

csv_files = [f for f in os.listdir(DATASET_DIR) if f.endswith(".csv")]
print("\nCSV file(s) found:", csv_files)
```
